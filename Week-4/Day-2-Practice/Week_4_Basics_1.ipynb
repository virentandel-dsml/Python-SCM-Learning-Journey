## 1. Load the Dataset

Load the supplied CSV dataset into a Pandas DataFrame.

## 4. Select Inventory Fields

Create a focused view containing the fields needed for inventory analysis.

```python
inventory_df = df[['Plant','Material','Opening_Inventory_Units','Material_Consumption_Units','Closing_Inventory_Units','Safety_Stock_Units']]
inventory_df
```

## 5. Identify Low-Stock Materials

Business rule: closing inventory below safety stock requires attention.

```python
low_stock_df = inventory_df[inventory_df['Closing_Inventory_Units'] < inventory_df['Safety_Stock_Units']]
low_stock_df
```

## 6. Calculate Inventory Gap

Inventory Gap = Safety Stock − Closing Inventory.

```python
low_stock_df['Inventory_Gap_Units'] = low_stock_df['Safety_Stock_Units'] - low_stock_df['Closing_Inventory_Units']
low_stock_df
```

## 7. Find the Largest Inventory Gap

Identify the maximum safety-stock shortfall.

```python
low_stock_df['Inventory_Gap_Units'].max()
```

## 8. Identify the Material With the Largest Gap

Locate the record corresponding to the largest observed gap.

```python
low_stock_df[low_stock_df['Inventory_Gap_Units'] == 75]
```

## 9. Calculate Inventory Coverage

Inventory Coverage % = Closing Inventory ÷ Safety Stock × 100.

```python
low_stock_df['Inventory_Coverage_%'] = low_stock_df['Closing_Inventory_Units']/low_stock_df['Safety_Stock_Units']*100
low_stock_df
```

## Business Takeaway

The analysis identifies materials below safety stock, quantifies the inventory gap, highlights the largest shortfall, and measures inventory coverage.
