This notebook applies Pandas to a manufacturing supply chain scenario. The analysis focuses on practical business questions rather than Python syntax alone.


```python
import pandas as pd
```

## 1. Load the Dataset

Load the supplied CSV dataset into a Pandas DataFrame.

```python
production_df = pd.read_csv("Week_4_Basics_2_Dataset.csv")
```

## 2. Inspect the Data

Review the first records and confirm the dataset dimensions.

```python
production_df.head()
```

```python
production_df.shape
```

## 3. Calculate Production Variance

Production Variance = Actual Production − Planned Production.

```python
production_df['Production_Variance_Units'] = (production_df['Actual_Production_Units'] - production_df['Planned_Production_Units'])
production_df
```

## 4. Calculate Production Variance %

Normalize production variance against the planned quantity.

```python
production_df['Production_Variance_%'] = (production_df['Production_Variance_Units'] / production_df['Planned_Production_Units'] * 100)
production_df
```

## 5. Aggregate Actual Production by Plant

Convert line-level production records into plant-level totals.

```python
plant_production = (production_df.groupby('Plant')['Actual_Production_Units'].sum())
plant_production
```

## 6. Aggregate Planned Production by Plant

```python
plant_plan = (production_df.groupby('Plant')['Planned_Production_Units'].sum())
plant_plan
```

## 7. Combine Planned and Actual Production

Use `concat()` to create a plant-level comparison table.

```python
plant_summary = pd.concat([plant_plan, plant_production],axis=1)
plant_summary
```

## 8. Calculate Production Variance

Production Variance = Actual Production − Planned Production.

```python
plant_summary['Production_Variance_Units'] = (plant_summary['Actual_Production_Units'] - plant_summary['Planned_Production_Units'])
plant_summary
```

## 9. Calculate Production Achievement %

Production Achievement % = Actual Production ÷ Planned Production × 100.

```python
plant_summary['Production_Achievement_%'] = (plant_summary['Actual_Production_Units'] / plant_summary['Planned_Production_Units'] * 100)
plant_summary
```

## 10. Calculate Production Achievement %

Production Achievement % = Actual Production ÷ Planned Production × 100.

```python
underperforming_plants = plant_summary[plant_summary['Production_Achievement_%'] < 95]
underperforming_plants
```

## 11. Calculate Production Achievement %

Production Achievement % = Actual Production ÷ Planned Production × 100.

```python
lowest_achievement = plant_summary['Production_Achievement_%'].min()
lowest_plant = plant_summary[plant_summary['Production_Achievement_%'] == lowest_achievement]
lowest_plant
```

## Business Takeaway

The analysis converts production-line data into plant-level performance measures, enabling comparison of planned production, actual production, variance, and achievement.
