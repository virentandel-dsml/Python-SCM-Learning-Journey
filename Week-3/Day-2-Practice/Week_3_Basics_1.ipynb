```python
inventory = [120,105,95,80,70]  #units
```

```python
import numpy as np
```

```python
inv = np.array(inventory)
print(inv)
print(f"Shape : {np.shape(inv)}")
print(f"Size : {np.size(inv)}")
print(f"Dimension : {np.ndim(inv)}")
```

```python
print(f"Inventory : {inv}")
print(f"Total Inventory : {np.sum(inv)}")
print(f"Average Inventory : {np.mean(inv)}")
```

```python
replenishment = np.array([10,10,10,10,10])  #units

new_inventory = inv + replenishment

print(f"Current Inventory : {inv}")
print(f"Replenishment : {replenishment}")
print(f"New Inventory : {new_inventory}")
```

```python
new_inventory = inv * 1.2   #increased replenishment by 20%
print(f"Current Inventory : {inv}")
print(f"Inventory after 20% increase : {new_inventory}")
```

```python
consumption = np.array([15,15,15,15,15])
post_consumption_inv = new_inventory - consumption
print(f"Current Inventory: {new_inventory} - Total Qty: {np.sum(new_inventory)}")
print(f"Inventory post consumption: {post_consumption_inv} - Total Qty: {np.sum(post_consumption_inv)}")
```

```python
suppliers = ["Apex Components","Vertex Industries","Nova Engineering","Prime Tech","Delta Manufacturing"]
lead_time = np.array([6,9,7,12,8])
```

```python
supplier = np.array(suppliers)
print(f"Total Suppliers: {np.size(suppliers)}")
print(f"Average Lead Time: {np.mean(lead_time)}")
print(f"Highest Lead Time: {np.max(lead_time)}")
print(f"Lowest Lead Time: {np.min(lead_time)}")
```

```python
target_lead_time = 8
```

```python
supplier_performance = lead_time - target_lead_time
print(f"Supplier Performance : {supplier_performance}")
```

```python
lead_time_reduction = 1
new_lead_time = lead_time - lead_time_reduction
print(f"Lead Time after improvement: {new_lead_time}")
```
