```python
import numpy as np
production = np.array([    #week wise production planning
    [500,550,600,580],     #Product A
    [300,320,350,340],     #Prodcut B
    [450,470,490,520]      #Product C
])
print(f"Production Plan: {production}")
print(f"Shape: {np.shape(production)}")
print(f"Dimension: {np.ndim(production)}")
```

```python
print(f"Total Production: {np.sum(production)}")
print(f"Average Production: {round(np.mean(production),2)}")
```

```python
print(f"Weekly production: {np.sum(production,axis=0)}")
```

```python
print(f"Product wise production per month: {np.sum(production, axis=1)}")
```

```python
new_production = 1.1*production
print(f"Current production: {production}")
print(f"Current total production: {np.sum(production)}\n")
print(f"Production after 10% increase: {new_production}")
print(f"Total production after 10% increase: {np.sum(new_production)}\n")
print(f"Total production increase: {np.sum(new_production)-np.sum(production)}")
```

```python
print(f"Product A's Week 3 production = {production[0,2]}")
```

```python
print(f"Product C's Week 2 production = {production[2,1]}")
```

```python
print(f"Week 2 and week 3 for all products: {production[:,1:3]}")
```

```python
capacity = np.array([       #production capacity
    [550, 550, 650, 600],
    [350, 350, 350, 350],
    [500, 500, 500, 550]
])
```

```python
print(f"Total Available Capacity: {np.sum(capacity) - np.sum(production)}")
```

```python
available_capacity = capacity - production
print(f"Capacity : {capacity}\n")
print(f"Planned Production: {production}\n")
print(f"Available Capacity: {available_capacity}")
```
