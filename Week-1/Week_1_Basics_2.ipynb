Industrial Bearing Inventory Value  
Calculate total inventory value using multiplication.


```python
product_name = "Industrial Bearing"
current_stock_ib = 180  # units
unit_cost_ib = 450      # INR per unit
safety_stock_ib = 150   # units
```

```python
inventory_value_ib = current_stock_ib * unit_cost_ib
print(inventory_value_ib)
```
