```python
warehouses = ['Pune', 'Mumbai', 'Chennai', 'Delhi']
current_stock = [250, 120, 300, 90]
safety_stock = [180, 150, 250, 100]
```

```python
print(f"{warehouses[0]} Warehouse:", end=" ")
if current_stock[0] > safety_stock[0]:
    print("Healthy")
else:
    print("Reorder Required")

print(f"{warehouses[1]} Warehouse:", end=" ")
if current_stock[1] > safety_stock[1]:
    print("Healthy")
else:
    print("Reorder Required")

print(f"{warehouses[2]} Warehouse:", end=" ")
if current_stock[2] > safety_stock[2]:
    print("Healthy")
else:
    print("Reorder Required")

print(f"{warehouses[3]} Warehouse:", end=" ")
if current_stock[3] > safety_stock[3]:
    print("Healthy")
else:
    print("Reorder Required")
```

Purchase Order Approval Logic  
Determine whether purchase orders require manager approval based on value thresholds.

```python
purchase_orders = ['P01001', 'P01002', 'P01003', 'P01004']
order_value = [75000, 120000, 45000, 210000]
```

```python
print(f"{purchase_orders[0]} : ", end="")
if order_value[0] > 100000:
    print("Manager Approval Required")
else:
    print("Auto Approved")

print(f"{purchase_orders[1]} : ", end="")
if order_value[1] > 100000:
    print("Manager Approval Required")
else:
    print("Auto Approved")

print(f"{purchase_orders[2]} : ", end="")
if order_value[2] > 100000:
    print("Manager Approval Required")
else:
    print("Auto Approved")

print(f"{purchase_orders[3]} : ", end="")
if order_value[3] > 100000:
    print("Manager Approval Required")
else:
    print("Auto Approved")
```

Shipment Tracking  
Check shipment delays and print delivery status.

```python
shipment_number = ["SH1001", "SH1002", "SH1003", "SH1004"]
delay_days = [0, 3, 7, 1]
```

```python
if delay_days[0] == 0:
    print(f"{shipment_number[0]} : On Time")
else:
    print(f"{shipment_number[0]} : Delayed by {delay_days[0]} days")

if delay_days[1] == 0:
    print(f"{shipment_number[1]} : On Time")
else:
    print(f"{shipment_number[1]} : Delayed by {delay_days[1]} days")

if delay_days[2] == 0:
    print(f"{shipment_number[2]} : On Time")
else:
    print(f"{shipment_number[2]} : Delayed by {delay_days[2]} days")

if delay_days[3] == 0:
    print(f"{shipment_number[3]} : On Time")
else:
    print(f"{shipment_number[3]} : Delayed by {delay_days[3]} days")
```

Product and Supplier Mapping  
Display product details, supplier name, and inventory health.

```python
products = ['Bearing', 'Motor', 'Shaft']
current_stock = [180, 90, 250]
safety_stock = [150, 120, 200]
suppliers = ['ABC Industries', 'XYZ Engineering', 'Techno Gear']
```

```python
print(f"Product : {products[0]}")
print(f"Supplier : {suppliers[0]}")
if current_stock[0] > safety_stock[0]:
    print("Status: Healthy")
else:
    print("Status: Reorder Required")

print(f"Product : {products[1]}")
print(f"Supplier : {suppliers[1]}")
if current_stock[1] > safety_stock[1]:
    print("Status: Healthy")
else:
    print("Status: Reorder Required")

print(f"Product : {products[2]}")
print(f"Supplier : {suppliers[2]}")
if current_stock[2] > safety_stock[2]:
    print("Status: Healthy")
else:
    print("Status: Reorder Required")
```
