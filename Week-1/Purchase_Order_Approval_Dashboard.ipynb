Mini Project 2: Purchase Order Approval Dashboard  

Objective:  
To automate purchase‑order approval decisions based on predefined thresholds, supplier details, and buyer information using Python conditionals and formatted output.


```python
po_number = ['PO1001', 'PO1002', 'PO1003', 'PO1004']  # Purchase order numbers
supplier = ['ABC Industries', 'XYZ Engineering', 'Techno Gear Industries', 'Prime Components']  # Supplier names
buyer = ['Soham', 'Mohan', 'Priya', 'Dhruti']  # Buyer names
order_value = [115000, 89000, 125000, 99500]  # Purchase order total value in INR
approval_threshold = 100000  # Threshold in INR for auto approval
```

Purchase Order Approval Dashboard  
Display PO details, supplier, buyer, and approval status based on value threshold.

```python
print("==============================================")
print("Purchase Order Approval Dashboard")
print("==============================================")
print("\n")

# PO 1
print(f"Purchase Order : {po_number[0]}")
print(f"Supplier Name : {supplier[0]}")
print(f"Buyer : {buyer[0]}")
print(f"Total PO Value : {order_value[0]} INR")
if order_value[0] < approval_threshold:
    print("Status : Auto Approved")
else:
    print("Status : Manager Approval Required")
print("\n----------------------------------------------\n")

# PO 2
print(f"Purchase Order : {po_number[1]}")
print(f"Supplier Name : {supplier[1]}")
print(f"Buyer : {buyer[1]}")
print(f"Total PO Value : {order_value[1]} INR")
if order_value[1] < approval_threshold:
    print("Status : Auto Approved")
else:
    print("Status : Manager Approval Required")
print("\n----------------------------------------------\n")

# PO 3
print(f"Purchase Order : {po_number[2]}")
print(f"Supplier Name : {supplier[2]}")
print(f"Buyer : {buyer[2]}")
print(f"Total PO Value : {order_value[2]} INR")
if order_value[2] < approval_threshold:
    print("Status : Auto Approved")
else:
    print("Status : Manager Approval Required")
print("\n----------------------------------------------\n")

# PO 4
print(f"Purchase Order : {po_number[3]}")
print(f"Supplier Name : {supplier[3]}")
print(f"Buyer : {buyer[3]}")
print(f"Total PO Value : {order_value[3]} INR")
if order_value[3] < approval_threshold:
    print("Status : Auto Approved")
else:
    print("Status : Manager Approval Required")
print("\n----------------------------------------------\n")
```
