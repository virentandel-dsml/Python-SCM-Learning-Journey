Mini Project 3: Customer Order Fulfillment Dashboard  

Objective:  
To build a Python‑based dashboard that checks available stock against customer orders and determines whether each order can be dispatched or requires replenishment.


```python
product = ['Bearing', 'Motor', 'Gear', 'Shaft']  # Items
customer = ['Tata Motors', 'Mahindra', 'Ashok Leyland', 'L&T']  # Customer names
available_stock = [180, 90, 260, 110]  # Units available
customer_order = [120, 100, 150, 80]   # Units ordered
```

Customer Order Fulfillment Dashboard  
Display customer‑wise details including product, available quantity, order quantity, and fulfillment status.


```python
print("==============================================")
print("Customer Order Fulfillment Dashboard")
print("==============================================")
print("\n")

# Customer 1
print(f"Customer : {customer[0]}")
print(f"Product : {product[0]}")
print(f"Available Quantity : {available_stock[0]}")
print(f"Order Quantity : {customer_order[0]}")
if available_stock[0] >= customer_order[0]:
    print("Status : Ready for Dispatch")
else:
    print("Status : Insufficient Inventory")
print("\n----------------------------------------------\n")

# Customer 2
print(f"Customer : {customer[1]}")
print(f"Product : {product[1]}")
print(f"Available Quantity : {available_stock[1]}")
print(f"Order Quantity : {customer_order[1]}")
if available_stock[1] >= customer_order[1]:
    print("Status : Ready for Dispatch")
else:
    print("Status : Insufficient Inventory")
print("\n----------------------------------------------\n")

# Customer 3
print(f"Customer : {customer[2]}")
print(f"Product : {product[2]}")
print(f"Available Quantity : {available_stock[2]}")
print(f"Order Quantity : {customer_order[2]}")
if available_stock[2] >= customer_order[2]:
    print("Status : Ready for Dispatch")
else:
    print("Status : Insufficient Inventory")
print("\n----------------------------------------------\n")

# Customer 4
print(f"Customer : {customer[3]}")
print(f"Product : {product[3]}")
print(f"Available Quantity : {available_stock[3]}")
print(f"Order Quantity : {customer_order[3]}")
if available_stock[3] >= customer_order[3]:
    print("Status : Ready for Dispatch")
else:
    print("Status : Insufficient Inventory")
print("\n----------------------------------------------\n")
```
