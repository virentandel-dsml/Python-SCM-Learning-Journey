Mini Project: Inventory Health Dashboard  

Objective:  
To build a Python‑based dashboard that evaluates inventory health across multiple warehouses and suppliers, calculates inventory value, and flags items that require reorder.


```python
products = ['Bearing', 'Motor', 'Shaft', 'Gear']        # Items
warehouse = ['Pune', 'Mumbai', 'Chennai', 'Delhi']      # Locations
suppliers = ['ABC Industries', 'XYZ Engineering', 'Techno Gear Industries', 'Prime Components']  # Supplier names
current_stock = [180, 90, 260, 110]                     # Units
safety_stock = [150, 120, 280, 100]                     # Units
unit_cost = [450, 7800, 1200, 2500]                     # INR per unit
```

```python
print("==============================================")
print("Inventory Health Dashboard")
print("==============================================")
print("\n")

# Product 1
print(f"Product : {products[0]}")
print(f"Warehouse : {warehouse[0]}")
print(f"Supplier : {suppliers[0]}")
print(f"Current Stock : {current_stock[0]}")
print(f"Safety Stock : {safety_stock[0]}")
print(f"Inventory Value : {current_stock[0] * unit_cost[0]} INR")

if current_stock[0] < safety_stock[0]:
    print("Status: Reorder Required")
else:
    print("Status: Healthy")

print("\n----------------------------------------------\n")

# Product 2
print(f"Product : {products[1]}")
print(f"Warehouse : {warehouse[1]}")
print(f"Supplier : {suppliers[1]}")
print(f"Current Stock : {current_stock[1]}")
print(f"Safety Stock : {safety_stock[1]}")
print(f"Inventory Value : {current_stock[1] * unit_cost[1]} INR")

if current_stock[1] < safety_stock[1]:
    print("Status: Reorder Required")
else:
    print("Status: Healthy")

print("\n----------------------------------------------\n")

# Product 3
print(f"Product : {products[2]}")
print(f"Warehouse : {warehouse[2]}")
print(f"Supplier : {suppliers[2]}")
print(f"Current Stock : {current_stock[2]}")
print(f"Safety Stock : {safety_stock[2]}")
print(f"Inventory Value : {current_stock[2] * unit_cost[2]} INR")

if current_stock[2] < safety_stock[2]:
    print("Status: Reorder Required")
else:
    print("Status: Healthy")

print("\n----------------------------------------------\n")

# Product 4
print(f"Product : {products[3]}")
print(f"Warehouse : {warehouse[3]}")
print(f"Supplier : {suppliers[3]}")
print(f"Current Stock : {current_stock[3]}")
print(f"Safety Stock : {safety_stock[3]}")
print(f"Inventory Value : {current_stock[3] * unit_cost[3]} INR")

if current_stock[3] < safety_stock[3]:
    print("Status: Reorder Required")
else:
    print("Status: Healthy")
print("\n----------------------------------------------\n")
```
