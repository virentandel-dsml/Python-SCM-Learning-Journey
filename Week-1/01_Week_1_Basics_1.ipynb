```python
current_stock = 420  # units
unit_cost = 1850     # INR per unit
transportation_cost = 12000  # INR
monthly_demand = 300  # units
```

```python
# Total Inventory Value
total_inventory_value = current_stock * unit_cost
print(total_inventory_value)
```

```python
# Inventory remaining after selling 150 units
remaining_stock = current_stock - 150
print(remaining_stock)
```

```python
# Purchase Cost for 80 New Units
purchase_cost = unit_cost * 80
print(purchase_cost)
```

```python
# Total inventory cost after adding transportation cost
total_inv_cost_w_trans = total_inventory_value + transportation_cost
print(total_inv_cost_w_trans)
```

```python
# Average weekly demand
avg_weekly_demand = monthly_demand / 4
print(avg_weekly_demand)
```
