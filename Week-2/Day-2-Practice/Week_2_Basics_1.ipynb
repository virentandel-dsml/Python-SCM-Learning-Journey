## 1. Working with Product Lists

Create a product list and practice basic iteration and indexing.


**Example 1 — Define the manufacturing products used throughout the examples.**


```python
products = ["Bearing","Motor","Gear","Shaft"]
```

**Example 2 — Loop through the product list and display each product.**


```python
for product in products:
  print(product)
```

**Example 3 — Use indexes to display products with sequential numbering.**


```python
for i in range(len(products)):
  print(f"{i+1}. {products[i]}")
```

## 2. Inventory Stock Assessment

Use conditional logic to evaluate stock levels and identify replenishment requirements.


**Example 4 — Define current inventory levels for each product.**


```python
stock = [120,45,80,30] #units
```

**Example 5 — Compare stock against a threshold to identify replenishment requirements.**


```python
for i in range(len(products)):
  print(f"{products[i]} => {stock[i]}")
```

**Example 6 — Classify inventory as healthy, monitor, or replenishment required.**


```python
for i in range(len(products)):
  if stock[i] > 50:
    print(f"{products[i]} => {stock[i]} => OK")
  else:
    print(f"{products[i]} => {stock[i]} => Replenishment Required")
```

**Example 7 — Calculate the quantity required to reach the target stock level.**


```python
for i in range(len(products)):
  if stock[i] < 50:
    print(f"{products[i]} => {stock[i]} => Replenishment Required")
  elif stock[i] < 100:
    print(f"{products[i]} => {stock[i]} => Monitor")
  else:
    print(f"{products[i]} => {stock[i]} => Healthy")
```

**Example 8 — Assign replenishment priority based on current stock levels.**


```python
for i in range(len(products)):
  if stock[i] < 100:
    print(f"{products[i]} => {stock[i]} => Replenish {100-stock[i]} units")
  else:
    print(f"{products[i]} => {stock[i]} => No Replenishment required")
```

## 3. Order Fulfillment Analysis

Compare customer demand with available inventory to assess fulfillment status.


**Example 9 — Define customer order quantities and available inventory.**


```python
for i in range(len(products)):
  if stock[i] < 50:
    print(f"{products[i]} => {stock[i]} => High Priority")
  elif stock[i] < 100:
    print(f"{products[i]} => {stock[i]} => Medium Priority")
  else:
    print(f"{products[i]} => {stock[i]} => No Replenishment Required")
```

**Example 10 — Determine whether each order is ready, partially fulfilled, or backordered.**


```python
products = ["Bearing","Motor","Gear","Shaft"]  #items
order_qty = [50,60,40,25]   #units
available_qty = [70,45,40,10]   #units
```

## 4. Inventory Count Variance

Compare system inventory with physical counts and classify quantity variances.


**Example 11 — Define system quantities and physical count quantities.**


```python
for i in range(len(products)):
  if available_qty[i] >= order_qty[i]:
    print(f"{products[i]} => Order: {order_qty[i]} => Stock: {available_qty[i]} => Order Ready")
  elif available_qty[i] > 0 and available_qty[i] < order_qty[i]:
    print(f"{products[i]} => Order: {order_qty[i]} => Stock: {available_qty[i]} => Partial Fulfillment")
  else:
    print(f"{products[i]} => Order: {order_qty[i]} => Stock: {available_qty[i]} => Backorder")
```

**Example 12 — Calculate inventory variance and classify the result.**


```python
products = ["Bearing","Motor","Gear","Shaft"]
system_qty = [100,50,80,40]  #unit
counted_qty = [98,50,75,44]  #unit
```

## 5. Supplier Lead Time Analysis

Classify suppliers according to lead-time thresholds for performance monitoring.


**Example 13 — Define suppliers and their respective lead times.**


```python
for i in range(len(products)):
  variance =  counted_qty[i] - system_qty[i]
  if variance == 0:
    print(f"{products[i]} => System: {system_qty[i]}, Counted: {counted_qty[i]}, Variance: {variance} => Matched")
  elif -5 <= variance <= 5:
    print(f"{products[i]} => System: {system_qty[i]}, Counted: {counted_qty[i]}, Variance: {variance} => Minor Variance")
  else:
    print(f"{products[i]} => System: {system_qty[i]}, Counted: {counted_qty[i]}, Variance: {variance} => Significant Variance")
```

**Example 14 — Classify suppliers using lead-time performance thresholds.**


```python
suppliers = ["Supplier A","Supplier B","Supplier C","Supplier D"]
lead_times = [4,9,6,12]
```

## 6. Order Priority Classification

Use business rules to classify orders according to requested quantities.


**Example 15 — Define customer orders and requested quantities.**


```python
for i, supplier in enumerate(suppliers):
  if lead_times[i] <=5:
    print(f"{suppliers[i]} => Lead Time: {lead_times[i]} days => Good")
  elif lead_times[i] <=10:
    print(f"{suppliers[i]} => Lead Time: {lead_times[i]} days => Monitor")
  else:
    print(f"{suppliers[i]} => Lead Time: {lead_times[i]} days => Critical")
```

**Example 16 — Assign order priority based on requested quantity.**


```python
orders =["ORD101","ORD102","ORD103","ORD104"]  #order_id
quantities = [25,8,40,15]  #units
```

## 7. Purchase Order Approval

Apply approval rules based on purchase order monetary value.


**Example 17 — Define purchase orders and their monetary values.**


```python
for i,order in enumerate(orders):
  if quantities[i] >= 30:
    print(f"{order} => Quantity: {quantities[i]} => High Priority")
  elif quantities[i] >= 10:
    print(f"{order} => Quantity: {quantities[i]} => Normal Priority")
  else:
    print(f"{order} => Quantity: {quantities[i]} => Low Priority")
```

**Example 18 — Apply approval rules based on purchase order value.**


```python
po_numbers = ["PO101","PO102","PO103","PO104"]  #purchase_order
po_values = [45000,120000,75000,200000]  #INR
```

## 8. Customer Shipment Status

Calculate pending quantities and classify customer shipment progress.


**Example 19 — Define customer orders and shipped quantities.**


```python
for i,po_number in enumerate(po_numbers):
  if po_values[i] < 50000:
    print(f"{po_number} => Value: {po_values[i]} => Auto Approval")
  elif po_values[i] < 100000:
    print(f"{po_number} => Value: {po_values[i]} => Manager Approval Required")
  else:
    print(f"{po_number} => Value: {po_values[i]} => Senior Approval Required")
```

**Example 20 — Calculate pending quantities and classify shipment status.**


```python
customers = ["Customer A","Customer B","Customer C","Customer D"]
ordered_qty = [100,80,150,60]  #units
shipped_qty = [100,50,120,0]  #units
```

**Example 21 — Review shipment status across customers using the same business logic.**


```python
for i,customer in enumerate(customers):
  pending_qty = ordered_qty[i]-shipped_qty[i]
  if  pending_qty == 0:
    print(f"{customer} => Ordered: {ordered_qty[i]} => Shipped: {shipped_qty[i]} => Pending: {pending_qty} => Fully Shipped")
  elif 0 < pending_qty < 50:
    print(f"{customer} => Ordered: {ordered_qty[i]} => Shipped: {shipped_qty[i]} => Pending: {pending_qty} => Partially Shipped")
  else:
    print(f"{customer} => Ordered: {ordered_qty[i]} => Shipped: {shipped_qty[i]} => Pending: {pending_qty} => Significant Pending")
```
