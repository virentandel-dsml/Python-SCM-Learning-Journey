## 1. Production Capacity Planning

Evaluate whether planned production quantities can be executed within available production-line capacity.


**Example 1 — Define production lines, products, available capacity, and the planned production quantities.**


```python
lines = ["Production Line 1","Production Line 2"]
products = ["Pump","Valve","Motor"]
capacity = [
    [100,80,60],
    [120,70,90]
    ]  #each row represents production line, and each column represents a product
production_plan = [
    [90,85,50],
    [100,60,95]
    ]  #each row represents production line, and each column represents a product
```

**Example 2 — Compare planned production against capacity for every production line and product combination.**


```python
for i,line in enumerate(lines):
  for j,product in enumerate(products):
    if production_plan[i][j] <= capacity[i][j]:
      print(f"{line} => {product} => Planned: {production_plan[i][j]} => Capacity: {capacity[i][j]} => Within Capacity")
    else:
      print(f"{line} => {product} => Planned: {production_plan[i][j]} => Capacity: {capacity[i][j]} => Capacity Exceeded")
```

## 2. Transportation Cost Analysis

Calculate transportation cost across routes and vehicle types, then classify cost efficiency.


**Example 3 — Define transportation routes, vehicle types, distances, and applicable cost-per-kilometre rates.**


```python
routes = ["Pune-Mumbai","Pune-Nashik","Pune-Surat"]
vehicles = ["Truck","Mini Truck"]
distance = [      #km
    [150,160],
    [210,220],
    [420,430]
]  #each row represents a route and each column represents a vehicle
rate_per_km = [     #INR
    [40,30],
    [40,30],
    [45,35]
]  #each row represents a route and each column represents a vehicle
```

**Example 4 — Calculate transportation cost for each route-vehicle combination and classify the result.**


```python
for i,route in enumerate(routes):
  for j,vehicle in enumerate(vehicles):
    transportation_cost = distance[i][j] * rate_per_km[i][j]   #INR
    if transportation_cost < 6000:
      print(f"{route} => {vehicle} => Distance: {distance[i][j]} km => Cost: {transportation_cost} => Cost Efficient")
    elif transportation_cost < 10000:
      print(f"{route} => {vehicle} => Distance: {distance[i][j]} km => Cost: {transportation_cost} => Moderate Cost")
    else:
      print(f"{route} => {vehicle} => Distance: {distance[i][j]} km => Cost: {transportation_cost} => High Cost")
```

## 3. Warehouse Space Utilization

Measure warehouse zone utilization and classify space usage against defined thresholds.


**Example 5 — Define warehouse zones along with their total capacity and occupied space.**


```python
zones = ["Zone A","Zone B","Zone C"]
capacity = [1000,1500,800]  #sq ft
used_space = [750,1350,720]  #sq ft
```

**Example 6 — Calculate utilization percentage and classify each warehouse zone.**


```python
for i,zone in enumerate(zones):
  utilization_percent = (used_space[i]/capacity[i])*100
  if utilization_percent < 70:
    print(f"{zone} => Capacity: {capacity[i]} => Used: {used_space[i]} => Utilization: {utilization_percent}% => Underutilized")
  elif utilization_percent < 90:
    print(f"{zone} => Capacity: {capacity[i]} => Used: {used_space[i]} => Utilization: {utilization_percent}% => Healthy")
  else:
    print(f"{zone} => Capacity: {capacity[i]} => Used: {used_space[i]} => Utilization: {utilization_percent}% => High Utilization")
```

## 4. Machine Availability Analysis

Calculate machine running hours and availability to assess operational performance.


**Example 7 — Define planned operating hours and downtime for each machine.**


```python
machines = ["Machine A","Machine B","Machine C"]
planned_hours = [8,10,12]  #hrs
downtime_hours = [1,3,0.5]  #hrs
```

**Example 8 — Calculate running hours and machine availability percentage, then classify performance.**


```python
for machine,planned_hour,downtime_hour in zip(machines,planned_hours,downtime_hours):
  running_hours = planned_hour - downtime_hour
  availability_percent = (running_hours/planned_hour)*100
  if availability_percent >= 95:
    print(f"{machine} => Planned: {planned_hour} hrs => Downtime: {downtime_hour} hrs => Availability: {round(availability_percent,1)}% => Excellent")
  elif availability_percent >= 85:
    print(f"{machine} => Planned: {planned_hour} hrs => Downtime: {downtime_hour} hrs => Availability: {round(availability_percent,1)}% => Acceptable")
  else:
    print(f"{machine} => Planned: {planned_hour} hrs => Downtime: {downtime_hour} hrs => Availability: {round(availability_percent,1)}% => Poor")
```

## 5. Demand Forecast Variance

Compare forecasted demand with actual demand and identify over- or under-forecasting.


**Example 9 — Define forecasted and actual demand quantities for each product.**


```python
products = ["Pump","Valve","Motor","Bearing"]
forecast = [100,80,120,150]  #units
actual = [110,72,120,140]  #units
```

**Example 10 — Calculate demand variance and classify each product as over forecast, under forecast, or matched.**


```python
for product,forecast_qty,actual_qty in zip(products,forecast,actual):
  variance = actual_qty - forecast_qty
  if variance > 0:
    print(f"{product} => Forecast: {forecast_qty} => Actual: {actual_qty} => Variance: {variance} => Over Forecast")
  elif variance < 0:
    print(f"{product} => Forecast: {forecast_qty} => Actual: {actual_qty} => Variance: {variance} => Under Forecast")
  else:
    print(f"{product} => Forecast: {forecast_qty} => Actual: {actual_qty} => Variance: {variance} => Matched")
```

## 6. Supplier Delivery Performance

Compare supplier-promised lead times with actual delivery times to identify delays.


**Example 11 — Define supplier-promised delivery times and actual delivery times.**


```python
suppliers = ["Supplier A","Supplier B","Supplier C","Supplier D"]
promised_days = [5,7,10,6]  #days
actual_days = [5,9,8,6]  #days
```

**Example 12 — Calculate delivery delay and classify suppliers as early, on time, or late.**


```python
for supplier,promised_day,actual_day in zip(suppliers,promised_days,actual_days):
  delay = actual_day - promised_day
  if delay == 0:
    print(f"{supplier} => Promised: {promised_day} days => Actual: {actual_day} days => Delay: {delay} days => On Time")
  elif delay > 0:
    print(f"{supplier} => Promised: {promised_day} days => Actual: {actual_day} days => Delay: {delay} days => Late")
  else:
    print(f"{supplier} => Promised: {promised_day} days => Actual: {actual_day} days => Delay: {delay} days => Early")
```

## 7. Warehouse Picking Accuracy

Calculate order-picking accuracy and classify fulfillment performance.


**Example 13 — Define customer order quantities and the quantities picked from the warehouse.**


```python
orders = ["ORD101","ORD102","ORD103","ORD104"]
ordered_qty = [100,80,120,50]  #units
picked_qty = [98,80,114,45]  #units
```

**Example 14 — Calculate picking accuracy and classify order-picking performance.**


```python
i=0
while i < len(orders):
  accuracy = (picked_qty[i]/ordered_qty[i])*100
  if accuracy >=98:
    print(f"{orders[i]} => Ordered: {ordered_qty[i]} => Picked: {picked_qty[i]} => Accuracy: {round(accuracy,1)}% => Excellent")
  elif accuracy >=95:
    print(f"{orders[i]} => Ordered: {ordered_qty[i]} => Picked: {picked_qty[i]} => Accuracy: {round(accuracy,1)}% => Good")
  else:
    print(f"{orders[i]} => Ordered: {ordered_qty[i]} => Picked: {picked_qty[i]} => Accuracy: {round(accuracy,1)}% => Needs Improvement")
  i += 1
```

## 8. Production Scrap Analysis

Calculate scrap rates and identify products requiring attention from a manufacturing-quality perspective.


**Example 15 — Define production quantities and scrap quantities for each product.**


```python
products = ["Pump","Valve","Motor","Bearing"]
produced_qty = [1000,800,1200,600]   #units
scrap_qty = [20,56,30,48]  #units
```

**Example 16 — Calculate scrap rate and classify manufacturing quality performance.**


```python
for i in range(len(products)):
  scrap_rate = (scrap_qty[i]/produced_qty[i])*100
  if scrap_rate <= 2:
    print(f"{products[i]} => Produced: {produced_qty[i]} => Scrap: {scrap_qty[i]} => Scrap Rate: {round(scrap_rate,1)} => Excellent")
  elif scrap_rate <=5:
    print(f"{products[i]} => Produced: {produced_qty[i]} => Scrap: {scrap_qty[i]} => Scrap Rate: {round(scrap_rate,1)} => Acceptable")
  else:
    print(f"{products[i]} => Produced: {produced_qty[i]} => Scrap: {scrap_qty[i]} => Scrap Rate: {round(scrap_rate)} => High Scrap")
```

## 9. Material Price Variance

Compare standard material prices with actual prices to identify purchasing price movements.


**Example 17 — Define standard and actual prices for key raw materials.**


```python
materials = ["Steel","Copper","Aluminium"]
standard_price = [100,250,180]  #INR per unit
actual_price = [105,240,180]  #INR per unit
```

**Example 18 — Calculate material price variance and identify price increases, decreases, or no change.**


```python
for material,standard_rate,actual_rate in zip(materials,standard_price,actual_price):
  price_variance = actual_rate-standard_rate
  if price_variance > 0:
    print(f"{material} => Standard: {standard_rate} => Actual: {actual_rate} => Variance = {price_variance} => Price Increase")
  elif price_variance < 0:
    print(f"{material} => Standard: {standard_rate} => Actual: {actual_rate} => Variance = {price_variance} => Price Decrease")
  else:
    print(f"{material} => Standard: {standard_rate} => Actual: {actual_rate} => Variance = {price_variance} => No Change")
```

## 10. Supplier Quality Performance

Calculate supplier defect rates and classify incoming material quality performance.


**Example 19 — Define received and defective quantities for each supplier.**


```python
suppliers = ["Supplier A","Supplier B","Supplier C","Supplier D"]
received_qty = [1000,800,1200,600]  #units
defective_qty = [10,32,18,48]  #units
```

**Example 20 — Calculate supplier defect rate and classify incoming quality performance.**


```python
i=0
while i < len(suppliers):
  defect_rate = defective_qty[i]/received_qty[i]*100
  if defect_rate <=2:
    print(f"{suppliers[i]} => Received: {received_qty[i]} => Defective: {defective_qty[i]} => Defect Rate: {round(defect_rate,1)}% => Excellent")
  elif defect_rate <=5:
    print(f"{suppliers[i]} => Received: {received_qty[i]} => Defective: {defective_qty[i]} => Defect Rate: {round(defect_rate,1)}% => Acceptable")
  else:
    print(f"{suppliers[i]} => Received: {received_qty[i]} => Defective: {defective_qty[i]} => Defect Rate: {round(defect_rate,1)}% => Poor")
  i += 1
```
