# Supplier Performance & Risk Assessment

This project evaluates supplier performance across delivery reliability, quality, and purchasing price variance. The analysis then combines these KPIs to identify supplier weaknesses and assess overall supplier risk.

### Business Objective
Identify suppliers that may be affecting Supply Chain performance and highlight where corrective action or closer monitoring may be required.

### KPIs Evaluated
- Delivery performance
- Defect rate
- Purchase price variance
- Overall supplier performance score
- Weakest supplier KPI
- Supplier risk classification

**Project setup:** Define supplier master data and the operational, quality, and pricing inputs used in the assessment.


```python
suppliers = ["Apex Components","Vertex Industrial","Nova Engineering","Prime Tech Manufacturing"]
promised_days = [5, 7, 10, 6]  #days
actual_days = [5, 9, 8, 7]  #days
received_qty = [1000, 800, 1200, 600]  #units
defective_qty = [10, 32, 18, 48]  #units
standard_price = [100, 250, 180, 150]  #INR per unit
actual_price = [105, 240, 195, 150]  #INR per unit
```

## 1. Calculate Supplier Performance KPIs

Calculate the core performance measures for each supplier using promised versus actual delivery time, received versus defective quantities, and standard versus actual purchase prices.

**KPI calculation:** Convert raw supplier data into measurable performance indicators.


```python
for supplier,promised_day,actual_day,rcv_qty,dft_qty,std_prc,act_prc in zip(suppliers,promised_days,actual_days,received_qty,defective_qty,standard_price,actual_price):
  delivery_delay = actual_day - promised_day
  defect_rate = dft_qty/rcv_qty*100
  price_variance = act_prc-std_prc
  if delivery_delay == 0:
    print(supplier)
    print(f"Delivery: On Time")
    print(f"Defect Rate: {round(defect_rate,1)}%")
    print(f"Price Variance: {price_variance}\n")
  elif delivery_delay > 0:
    print(supplier)
    print(f"Delivery: Late")
    print(f"Defect Rate: {round(defect_rate,1)}%")
    print(f"Price Variance: {price_variance}\n")
  else:
    print(supplier)
    print(f"Delivery: Early")
    print(f"Defect Rate: {round(defect_rate,1)}%")
    print(f"Price Variance: {price_variance}\n")
```

## 2. Supplier Performance Rating

Translate the individual KPI results into supplier-level performance ratings using the business rules defined in the project.

**Performance rating:** Apply the project's rating rules to evaluate individual supplier performance.


```python
print("===================================")
print("Supplier Performance Ratings")
print("===================================\n")
for supplier,promised_day,actual_day,rcv_qty,dft_qty,std_prc,act_prc in zip(suppliers,promised_days,actual_days,received_qty,defective_qty,standard_price,actual_price):
  delivery_delay = actual_day - promised_day
  if delivery_delay == 0:
    print(supplier)
    print(f"Delivery: Good")
  elif delivery_delay <= 2:
    print(supplier)
    print(f"Delivery: Monitor")
  else:
    print(supplier)
    print(f"Delivery: Poor")

  defect_rate = dft_qty/rcv_qty*100
  if defect_rate <=2:
    print(f"Quality: Good")
  elif defect_rate <=5:
    print(f"Quality: Monitor")
  else:
    print(f"Quality: Poor")

  price_variance = act_prc-std_prc
  if price_variance <=0:
    print(f"Price: Good\n")
  elif price_variance <=10:
    print(f"Price: Monitor\n")
  else:
    print(f"Price: Poor\n")
```

## 3. Overall Supplier Score

Combine the supplier performance measures to create an overall score that supports comparison across suppliers.

**Score aggregation:** Combine KPI results into an overall supplier score.


```python
print("===================================")
print("Overall Supplier Score")
print("===================================\n")
for supplier,promised_day,actual_day,rcv_qty,dft_qty,std_prc,act_prc in zip(suppliers,promised_days,actual_days,received_qty,defective_qty,standard_price,actual_price):
  delivery_delay = actual_day - promised_day
  if delivery_delay == 0:
    delivery_delay_score = 3
  elif delivery_delay <= 2:
    delivery_delay_score = 2
  else:
    delivery_delay_score = 1

  defect_rate = dft_qty/rcv_qty*100
  if defect_rate <=2:
    defect_rate_score = 3
  elif defect_rate <=5:
    defect_rate_score = 2
  else:
    defect_rate_score = 1

  price_variance = act_prc-std_prc
  if price_variance <=0:
    price_variance_score = 3
  elif price_variance <=10:
    price_variance_score = 2
  else:
    price_variance_score = 1

  overall_score = delivery_delay_score + defect_rate_score + price_variance_score

  print(f"Supplier: {supplier}")
  print(f"Delivery Score: {delivery_delay_score}")
  print(f"Quality Score: {defect_rate_score}")
  print(f"Price Score: {price_variance_score}")

  if overall_score >=8:
    print(f"Overall Score: {overall_score} - Strong Supplier\n")
  elif overall_score >=6:
    print(f"Overall Score: {overall_score} - Moderate Risk\n")
  else:
    print(f"Overall Score: {overall_score} - High Risk\n")
```

## 4. Supplier Score Review

Review the overall supplier scoring logic and resulting supplier ratings to identify relative performance differences.

**Score validation:** Review the scoring output across suppliers using the same assessment logic.


```python
print("===================================")
print("Overall Supplier Score")
print("===================================\n")
for supplier,promised_day,actual_day,rcv_qty,dft_qty,std_prc,act_prc in zip(suppliers,promised_days,actual_days,received_qty,defective_qty,standard_price,actual_price):
  delivery_delay = actual_day - promised_day
  if delivery_delay == 0:
    delivery_delay_score = 3
  elif delivery_delay <= 2:
    delivery_delay_score = 2
  else:
    delivery_delay_score = 1

  defect_rate = dft_qty/rcv_qty*100
  if defect_rate <=2:
    defect_rate_score = 3
  elif defect_rate <=5:
    defect_rate_score = 2
  else:
    defect_rate_score = 1

  price_variance = act_prc-std_prc
  if price_variance <=0:
    price_variance_score = 3
  elif price_variance <=10:
    price_variance_score = 2
  else:
    price_variance_score = 1

  overall_score = delivery_delay_score + defect_rate_score + price_variance_score

  print(f"Supplier: {supplier}")
  if overall_score >=8:
    print(f"Overall Score: {overall_score}")
    print(f"Assessment: Strong Supplier")
    print(f"Action: Continue / Preferred Supplier\n")
  elif overall_score >=6:
    print(f"Overall Score: {overall_score}")
    print(f"Assessment: Moderate Risk")
    print(f"Action: Monitor Performance\n")
  else:
    print(f"Overall Score: {overall_score}")
    print(f"Assessment: High Risk")
    print(f"Action: Supplier Improvement Required\n")
```

## 5. Identify the Weakest KPI

Determine which performance dimension represents the biggest weakness for each supplier, helping focus supplier improvement discussions.

**Weakest-area analysis:** Identify the KPI where each supplier performs least effectively.


```python
print("==============================")
print("Weakest KPI")
print("==============================")

for supplier, promised_day, actual_day, rcv_qty, dft_qty, std_prc, act_prc in zip(
    suppliers, promised_days, actual_days, received_qty,
    defective_qty, standard_price, actual_price
):

    # Delivery Score
    delivery_delay = actual_day - promised_day

    if delivery_delay == 0:
        delivery_delay_score = 3
    elif delivery_delay <= 2:
        delivery_delay_score = 2
    else:
        delivery_delay_score = 1

    # Quality Score
    defect_rate = dft_qty / rcv_qty * 100

    if defect_rate <= 2:
        defect_rate_score = 3
    elif defect_rate <= 5:
        defect_rate_score = 2
    else:
        defect_rate_score = 1

    # Price Score
    price_variance = act_prc - std_prc

    if price_variance <= 0:
        price_variance_score = 3
    elif price_variance <= 10:
        price_variance_score = 2
    else:
        price_variance_score = 1

    # Identify weakest KPI
    weakest_score = min(
        delivery_delay_score,
        defect_rate_score,
        price_variance_score
    )

    if delivery_delay_score == weakest_score:
        risk_driver = "Delivery"
    elif defect_rate_score == weakest_score:
        risk_driver = "Quality"
    else:
        risk_driver = "Price"

    print(f"\nSupplier: {supplier}")
    print(f"Delivery Score: {delivery_delay_score}")
    print(f"Quality Score: {defect_rate_score}")
    print(f"Price Score: {price_variance_score}")
    print(f"Risk Driver: {risk_driver}")
```

## 6. Supplier Risk Assessment

Use the supplier performance results to classify overall supplier risk and identify suppliers requiring greater management attention.

**Risk classification:** Translate supplier performance into an overall risk assessment.


```python
print("==============================")
print("Supplier Risk Assessment")
print("==============================")

for supplier, promised_day, actual_day, rcv_qty, dft_qty, std_prc, act_prc in zip(
    suppliers, promised_days, actual_days, received_qty,
    defective_qty, standard_price, actual_price
):

    # Delivery Score
    delivery_delay = actual_day - promised_day

    if delivery_delay == 0:
        delivery_delay_score = 3
    elif delivery_delay <= 2:
        delivery_delay_score = 2
    else:
        delivery_delay_score = 1

    # Quality Score
    defect_rate = dft_qty / rcv_qty * 100

    if defect_rate <= 2:
        defect_rate_score = 3
    elif defect_rate <= 5:
        defect_rate_score = 2
    else:
        defect_rate_score = 1

    # Price Score
    price_variance = act_prc - std_prc

    if price_variance <= 0:
        price_variance_score = 3
    elif price_variance <= 10:
        price_variance_score = 2
    else:
        price_variance_score = 1

    # Overall Score
    overall_score = (
        delivery_delay_score
        + defect_rate_score
        + price_variance_score
    )

    # Risk Driver
    weakest_score = min(
        delivery_delay_score,
        defect_rate_score,
        price_variance_score
    )

    if delivery_delay_score == weakest_score:
        risk_driver = "Delivery"
    elif defect_rate_score == weakest_score:
        risk_driver = "Quality"
    else:
        risk_driver = "Price"

    # Assessment + Recommended Action
    if overall_score >= 8:
        assessment = "Strong Supplier"
        action = "Continue / Preferred Supplier"

    elif overall_score >= 6:
        assessment = "Moderate Risk"
        action = f"Monitor {risk_driver} Performance"

    else:
        assessment = "High Risk"
        action = f"Supplier Improvement Required - {risk_driver}"

    # Final Output
    print(f"\nSupplier: {supplier}")
    print(f"Overall Score: {overall_score}")
    print(f"Risk Driver: {risk_driver}")
    print(f"Assessment: {assessment}")
    print(f"Recommended Action: {action}")
```

## Key Takeaway
